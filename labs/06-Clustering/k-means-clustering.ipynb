## Unsupervised Learning

### Clustering with K-means

Thực hành phân cụm dữ liệu Iris bằng K-means.


```python
# Import iris dataset
from sklearn.datasets import load_iris
# import k-Means clustering model
from sklearn.cluster import KMeans 
from sklearn.metrics import silhouette_score
```

```python
# Load the iris dataset
iris = load_iris()
X = iris.data
y = iris.target
```

```python
print(y)
```

```python
# Set k = 3 and fit the model

km = KMeans(n_clusters = 3, 
init = 'k-means++', max_iter = 300, n_init = 10, random_state = 0)
y_kmeans = km.fit_predict(X)
# Get the cluster labels
labels = km.labels_
# Evaluate the model using the silhouette score
silhouette_avg = silhouette_score(X, labels)
print("For n_clusters =", 3,
      "The average silhouette_score is :", silhouette_avg)

from sklearn.metrics import adjusted_rand_score

print(f"Adjusted Rand index: {adjusted_rand_score(y, labels):.4f}")

# Each cluster is assigned its most frequent species for interpretation.
import numpy as np
matched_species = np.empty_like(y)
for cluster in np.unique(labels):
    members = labels == cluster
    species = np.bincount(y[members]).argmax()
    matched_species[members] = species
    print(f"Cluster {cluster}: {members.sum()} samples, majority species = {iris.target_names[species]}")
print(f"Cluster purity: {np.mean(matched_species == y):.2%}")
```
