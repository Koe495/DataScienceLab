# Nhận diện chữ số viết tay bằng CNN

Thực hành tiền xử lý MNIST, huấn luyện CNN, chọn checkpoint theo validation loss và đánh giá trên tập test.

```python
%matplotlib inline
import matplotlib.pyplot as plt
from model import load_splits, train_model

EPOCHS = 5
train, validation, test = load_splits()
for name, (images, labels) in zip(('Train', 'Validation', 'Test'), (train, validation, test)):
    print(f'{name}: {images.shape}, {labels.shape}')
```

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 4), layout='constrained')
for i, ax in enumerate(axes.flat):
    ax.imshow(train[0][i, :, :, 0], cmap='gray')
    ax.set_title(str(train[1][i]))
    ax.axis('off')
plt.show()
```

## Huấn luyện và chọn checkpoint

Mô hình nhận ảnh 28 × 28 × 1 và trả về xác suất của 10 chữ số. Tập validation được tách từ dữ liệu train. Chỉ checkpoint có validation loss thấp nhất được dùng để đánh giá trên tập test.

```python
best_model, history = train_model(train, validation, epochs=EPOCHS)
best_model.summary()

fig, ax = plt.subplots(figsize=(7, 4), layout='constrained')
epoch = range(1, len(history.history['loss']) + 1)
ax.plot(epoch, history.history['loss'], label='Train')
ax.plot(epoch, history.history['val_loss'], label='Validation')
ax.set(xlabel='Epoch', ylabel='Cross-entropy loss')
ax.legend()
plt.show()
```

```python
metrics = best_model.evaluate(*test, verbose=0, return_dict=True)
print(f"Test accuracy: {metrics['accuracy']:.2%}")
print(f"Test loss: {metrics['loss']:.4f}")

probabilities = best_model.predict(test[0][:10], verbose=0)
fig, axes = plt.subplots(2, 5, figsize=(10, 4), layout='constrained')
for i, ax in enumerate(axes.flat):
    ax.imshow(test[0][i, :, :, 0], cmap='gray')
    ax.set_title(f'Đúng: {test[1][i]} · Dự đoán: {probabilities[i].argmax()}', fontsize=9)
    ax.axis('off')
plt.show()
```
